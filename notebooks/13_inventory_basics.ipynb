```python
import pandas as pd
import numpy as np
from pathlib import Path

data_path = Path("../data/processed/store_daily_sales.csv")

sales = pd.read_csv(data_path, parse_dates=["Date"])
sales = sales.sort_values(["Store", "Date"]).reset_index(drop=True)

print("Dataset shape:", sales.shape)
print("Stores:", sales["Store"].nunique())
print("Date range:", sales["Date"].min(), "to", sales["Date"].max())
display(sales.head())
```

```python
# Estimate average daily sales for each store
store_demand = (
    sales.groupby("Store")
    .agg(
        Average_Daily_Sales=("Sales", "mean"),
        Daily_Sales_Std=("Sales", "std"),
        Observed_Days=("Date", "nunique")
    )
    .reset_index()
)

display(store_demand.head(10))
```

```python
store_demand.to_csv(
    "../data/processed/day15_store_demand_baseline.csv",
    index=False
)

print("Day 15 demand baseline saved!")
```
