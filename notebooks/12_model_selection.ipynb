```python
from pathlib import Path
import pandas as pd

data_dir = Path("../data/processed")

rf = pd.read_csv(data_dir / "day9_random_forest_evaluation.csv")
xgb = pd.read_csv(data_dir / "day10_xgboost_evaluation.csv")

comparison = pd.concat([rf, xgb], ignore_index=True)
comparison = comparison.sort_values("MAE").reset_index(drop=True)

comparison.to_csv(
    data_dir / "day14_model_selection.csv",
    index=False
)

display(comparison)
```

### Day 14 — Model Selection

- Compared Random Forest and XGBoost using the same temporal validation setup.
- XGBoost achieved an MAE of 697.11, RMSE of 999.25, and sMAPE of 10.32%.
- Random Forest achieved an MAE of 710.63, RMSE of 1036.48, and sMAPE of 10.51%.
- XGBoost performed better across all three evaluation metrics on this holdout.
- XGBoost is selected as the current candidate forecasting model, subject to further validation and operational testing.
- Store-level errors should continue to be monitored because aggregate metrics may hide poor performance at individual stores.
